## Data and methods

Inspect source labels, eligibility, and matrix scale. GEO reports GC-RMA normalized log2 signals.

```python
from pathlib import Path
from src.data_loading import parse_geo
from src.preprocessing import select_samples
expression, all_metadata = parse_geo(Path('data/raw/GSE45827_series_matrix.txt.gz'))
selected, metadata = select_samples(expression, all_metadata)
print(expression.shape, selected.shape)
print(all_metadata.subtype.value_counts())
print(metadata.groupby(['subtype', 'batch']).size())
print(selected.stack().describe())
```
